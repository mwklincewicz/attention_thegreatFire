```python
df = pd.read_csv("../Csv Files/F_Merged.csv")
df.head()
```

```python
#Dictionaries containing the counters for conditions and ages
conditions = {"CM1_M":0,"CM2_M":0,"IM1_M":0, "IM2_M":0,"CM1_F":0,"CM2_F":0,"IM1_F":0, "IM2_F":0}
ages = {"CM1_M":0,"CM2_M":0,"IM1_M":0, "IM2_M":0,"CM1_F":0,"CM2_F":0,"IM1_F":0, "IM2_F":0}
count = 0

#Finding the minimum and maximum age
min = 100
max = 0

total = 0

for _,row in df.iterrows():

    if row["Age"] > max:
        max = row["Age"]

    if row["Age"] < min:
        min = row["Age"]


    if count == 2:
        total += row["Age"]
        if row["condition"] == 0:
            if row["Gender"] == "Male":
                conditions["CM1_M"] += 1
                ages["CM1_M"] += row["Age"]
            else:
                conditions["CM1_F"] += 1
                ages["CM1_F"] += row["Age"]
        elif row["condition"] == 1:
            if row["Gender"] == "Male":
                conditions["CM2_M"] += 1
                ages["CM2_M"] += row["Age"]
            else:
                conditions["CM2_F"] += 1
                ages["CM2_F"] += row["Age"]
        elif row["condition"] == 2:
            if row["Gender"] == "Male":
                conditions["IM1_M"] += 1
                ages["IM1_M"] += row["Age"]
            else:
                conditions["IM1_F"] += 1
                ages["IM1_F"] += row["Age"]
        else:
            if row["Gender"] == "Male":
                conditions["IM2_M"] += 1
                ages["IM2_M"] += row["Age"]
            else:
                conditions["IM2_F"] += 1
                ages["IM2_F"] += row["Age"]
        count=0
    else:
        count +=1


#Printing the Amount of people of each Gender in 
print(f"Condition\tMale\tFemale\tTotal\t%Female")
print(f"CM1\t\t{conditions["CM1_M"]}\t{conditions["CM1_F"]}\t{conditions["CM1_M"] + conditions["CM1_F"]}\t{(conditions["CM1_F"]/(conditions["CM1_M"] + conditions["CM1_F"]))*100}")
print(f"CM2\t\t{conditions["CM2_M"]}\t{conditions["CM2_F"]}\t{conditions["CM2_M"] + conditions["CM2_F"]}\t{(conditions["CM2_F"]/(conditions["CM2_M"] + conditions["CM2_F"]))*100}")
print(f"IM1\t\t{conditions["IM1_M"]}\t{conditions["IM1_F"]}\t{conditions["IM1_M"] + conditions["IM1_F"]}\t{(conditions["IM1_F"]/(conditions["IM1_M"] + conditions["IM1_F"]))*100}")
print(f"IM2\t\t{conditions["IM2_M"]}\t{conditions["IM2_F"]}\t{conditions["IM2_M"] + conditions["IM2_F"]}\t{(conditions["IM2_F"]/(conditions["IM2_M"] + conditions["IM2_F"]))*100:.1f}")

print("\n")
print(f"Condition\tMale\tFemale\tTotal")
print(f"CM1\t\t{ages["CM1_M"] / conditions["CM1_M"]:.2f}\t{ages["CM1_F"] / conditions["CM1_F"]:.2f}\t{(ages["CM1_M"] + ages["CM1_F"])/(conditions["CM1_M"] + conditions["CM1_F"])}")
print(f"CM2\t\t{ages["CM2_M"] / conditions["CM2_M"]:.2f}\t{ages["CM2_F"] / conditions["CM2_F"]:.2f}\t{(ages["CM2_M"] + ages["CM2_F"])/(conditions["CM2_M"] + conditions["CM2_F"])}")
print(f"IM1\t\t{ages["IM1_M"] / conditions["IM1_M"]:.2f}\t{ages["IM1_F"] / conditions["IM1_F"]:.2f}\t{(ages["IM1_M"] + ages["IM1_F"])/(conditions["IM1_M"] + conditions["IM1_F"])}")
print(f"IM2\t\t{ages["IM2_M"] / conditions["IM2_M"]:.2f}\t{ages["IM2_F"] / conditions["IM2_F"]:.2f}\t{(ages["IM2_M"] + ages["IM2_F"])/(conditions["IM2_M"] + conditions["IM2_F"]):.2f}")
```
